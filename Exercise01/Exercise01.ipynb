## ---------- EXCERCISE 2 -----------

```python
import cv2
```

```python
image = cv2.imread("images/nature.jpg")
```

```python
grayImage = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
```

```python
grayImage = cv2.flip(grayImage, 1)
```

```python
grayImage = cv2.resize(grayImage, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_AREA)
```

```python
cv2.imshow("gray, flipped and resized nature", grayImage)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## ---------- EXCERCISE 3 -----------

```python
import cv2
```

```python
image = cv2.imread("images/nature.jpg")
```

```python
image = cv2.rectangle(image, (100, 100), (250, 200), (255, 20, 10), 2)
```

```python
image = cv2.circle(image, center=(500, 200), radius=40, color=(0, 255, 30), thickness=3)
```

```python
image = cv2.line(image, (50, 400), (600, 50), (255, 20, 0), 5)
```

```python
image = cv2.putText(image, "Gabriel", (40, 40), cv2.FONT_HERSHEY_DUPLEX, 2, (20, 255, 0), 3)
```

```python
cv2.imshow("Shapes image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## ---------- EXCERCISE 4 -----------

```python
cv2.imwrite("images/shapeImage.jpg", image)
```

## ---------- EXCERCISE 5 -----------

```python
camera = cv2.VideoCapture(0)
```

### Loop is exited by pressing esc

```python
shouldRun = True

while(shouldRun):
    result, frame = camera.read()
    cv2.imshow("Webcam view", frame)
    
    key = cv2.waitKey(1)
    
    if key == 27: #esc key
        shouldRun = False
        
    elif key == 115: #s key
        cv2.imwrite("images/webcamCapture.png", frame)
        
camera.release()
cv2.destroyAllWindows()
```
